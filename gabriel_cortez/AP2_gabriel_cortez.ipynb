**Objetivo:** Desenvolver um modelo de classificação para predizer gênero a partir de textos de reviews.

## 1. Exploração dos Dados

```python
import pandas as pd

# Carregar dados
df_train = pd.read_csv('treino.csv')
df_val = pd.read_csv('validacao.csv')

print(f'Treino: {len(df_train)} reviews')
print(f'Validação: {len(df_val)} reviews')

print('
Distribuição de gênero (treino):')
print(df_train['reviewer_gender'].value_counts(dropna=False))

print('
Amostra de registros:')
print(df_train[['review_text','reviewer_gender']].head(5))
```

## 2. Pré-processamento

```python
import re
import numpy as np

def preprocessar(texto):
    if pd.isna(texto):
        return ''
    s = str(texto)
    s = s.lower()
    s = re.sub(r'http+|www+', ' URL ', s)
    s = re.sub(r'<.*?>', ' ', s)
    s = re.sub(r
, ' ', s)
    s = re.sub(r'+', ' NUM ', s)
    s = re.sub(r'+', ' ', s).strip()
    return s

# Aplicar pré-processamento
X_train_clean = df_train['review_text'].fillna('').apply(preprocessar)
y_train = df_train['reviewer_gender']

X_val_clean = df_val['review_text'].fillna('').apply(preprocessar)
y_val = df_val['reviewer_gender']

print('Exemplo pré-processado:')
print(X_train_clean.head(3))
```

## 3. Representação Textual (TF-IDF)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(ngram_range=(1,2), max_features=20000, min_df=5, sublinear_tf=True)

X_train_vec = vectorizer.fit_transform(X_train_clean)
X_val_vec = vectorizer.transform(X_val_clean)

print('Dimensão treino:', X_train_vec.shape)
print('Dimensão validação:', X_val_vec.shape)
```

## 4. Treinamento

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=2000, class_weight='balanced', C=0.5, random_state=42)

model.fit(X_train_vec, y_train)
print('Treinamento concluído')
```

## 5. Validação e Métricas

```python
from sklearn.metrics import f1_score, confusion_matrix, classification_report

y_pred = model.predict(X_val_vec)
f1 = f1_score(y_val, y_pred, average='weighted')
print(f'F1 Score: {f1:.4f}')

print('
Matriz de Confusão:')
print(confusion_matrix(y_val, y_pred))

print('
Relatório de Classificação:')
print(classification_report(y_val, y_pred, digits=4))
```

## 6. Salvar Modelo

```python
import pickle

# Salva junto o vetor de features e o classificador
artefatos = {'vectorizer': vectorizer, 'model': model}
with open('modelo_seu_nome.pkl', 'wb') as f:
    pickle.dump(artefatos, f, protocol=pickle.HIGHEST_PROTOCOL)

print('✓ Modelo salvo: modelo_seu_nome.pkl')
```
